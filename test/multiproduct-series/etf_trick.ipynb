### Etf trick

```python
import pandas as pd
import numpy as np

# Ejemplo con dos futuros: long y short
data = {
    'p1': [100, 101, 103, 104],
    'p2': [99, 100, 101, 102],
    'd1': [0, 0, 0, 0],
    'd2': [0, 0, 0, 0]
}
df = pd.DataFrame(data)
df['K'] = np.nan
df.loc[0, 'K'] = 1  # inversión inicial de $1

omega = np.array([1, -1])  # pesos: long uno, short otro
phi = np.array([1, 1])     # valor por punto

for t in range(1, len(df)):
    delta = np.array([
        df.loc[t, 'p1'] - df.loc[t-1, 'p1'],
        df.loc[t, 'p2'] - df.loc[t-1, 'p2']
    ])
    df.loc[t, 'K'] = df.loc[t-1, 'K'] + (omega * phi * delta).sum()

df
```
